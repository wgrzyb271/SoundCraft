```python
import os
import time
import subprocess
import pandas as pd
import soundfile as sf

DRIVE_DIR = "/content/drive/MyDrive/Spleeter_Benchmark"
OUTPUT_DIR = os.path.join(DRIVE_DIR, "audio")
CSV_PATH = os.path.join(DRIVE_DIR, "spleeter_benchmark_full.csv")
os.makedirs(OUTPUT_DIR, exist_ok=True)

if os.path.exists(CSV_PATH):
    df_done = pd.read_csv(CSV_PATH)
    done_count = len(df_done)
    print(f"Znaleziono {done_count} przetworzonych utworów. Wznawiam od {done_count + 1}.")
else:
    done_count = 0
    print("Brak postępu – zaczynam od zera.")

test_tracks = [track for track in mus if track.subset == 'test']
test_tracks = test_tracks[done_count:]
print(f"Do przetworzenia: {len(test_tracks)} utworów.")

for i, track in enumerate(test_tracks):
    global_index = done_count + i
    print(f"\n[{global_index + 1}/50] {track.name}")

    mixture_path = f"/content/mixture_{global_index}.wav"
    sf.write(mixture_path, track.audio, track.rate)

    start_time = time.time()
    result = subprocess.run(
        ["conda", "run", "-n", "spleeter_env", "python", "-m", "spleeter",
         "separate", "-p", "spleeter:4stems",
         "-o", OUTPUT_DIR, mixture_path],
        capture_output=True, text=True
    )
    inference_time = time.time() - start_time

    expected_folder = os.path.join(OUTPUT_DIR, f"mixture_{global_index}")
    if result.returncode != 0 or not os.path.exists(expected_folder):
        print(f"   ❌ BŁĄD – pomijam ten utwór.")
        print(f"   STDERR: {result.stderr[:300]}")
        if os.path.exists(mixture_path):
            os.remove(mixture_path)
        continue

    print(f"   ✅ OK – {inference_time:.2f} s")

    df = pd.DataFrame([{
        "model": "Spleeter",
        "track": track.name,
        "inference_time_sec": inference_time
    }])
    df.to_csv(CSV_PATH, mode='a', header=not os.path.exists(CSV_PATH), index=False)

    if os.path.exists(mixture_path):
        os.remove(mixture_path)

print("\n✅ Benchmark zakończony.")
```

```python
import os
import soundfile as sf
import musdb

REFERENCE_DIR = "/content/drive/MyDrive/Spleeter_Benchmark/reference"
os.makedirs(REFERENCE_DIR, exist_ok=True)

mus = musdb.DB(root="/content/musdb18", is_wav=False)
test_tracks = [t for t in mus if t.subset == 'test']

for track in test_tracks:
    track_dir = os.path.join(REFERENCE_DIR, track.name)
    os.makedirs(track_dir, exist_ok=True)
    for stem in ['vocals', 'drums', 'bass', 'other']:
        if stem in track.targets:
            audio = track.targets[stem].audio
            sf.write(os.path.join(track_dir, f"{stem}.wav"), audio, track.rate)
    print(f"Zapisano referencje dla: {track.name}")
```

```python
import os
import shutil
import musdb

OUTPUTS_DIR = "/content/drive/MyDrive/Spleeter_Benchmark/outputs"
SOURCE_DIR = "/content/drive/MyDrive/Spleeter_Benchmark/audio"
os.makedirs(OUTPUTS_DIR, exist_ok=True)

mus = musdb.DB(root="/content/musdb18", is_wav=False)
test_tracks = [t for t in mus if t.subset == 'test']

for i, track in enumerate(test_tracks):
    src = os.path.join(SOURCE_DIR, f"mixture_{i}")
    dst = os.path.join(OUTPUTS_DIR, track.name)
    if os.path.exists(src):
        if os.path.exists(dst):
            shutil.rmtree(dst)
        shutil.copytree(src, dst)
        print(f"Skopiowano mixture_{i} -> {track.name}")
    else:
        print(f"⚠️ Brak mixture_{i} dla {track.name}")
```

```python
import csv
import numpy as np
import soundfile as sf
import torch
from pathlib import Path
from audiobox_aesthetics.infer import initialize_predictor

RESULTS_CSV = "/content/drive/MyDrive/Spleeter_Benchmark/results.csv"
OUTPUTS_DIR = Path("/content/drive/MyDrive/Spleeter_Benchmark/outputs")

print("Ładowanie audiobox_aesthetics...")
predictor = initialize_predictor()
print("Model gotowy.")

with open(RESULTS_CSV, "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    rows = list(reader)

print(f"Wczytano {len(rows)} wierszy.")

for i, row in enumerate(rows):
    if row.get("PQ") not in ("", None, "not_supported"):
        continue

    track = row["track"]
    stem = row["stem"]
    est_path = OUTPUTS_DIR / track / f"{stem}.wav"

    if not est_path.exists():
        continue

    try:
        audio, sr = sf.read(str(est_path), always_2d=True, dtype="float32")
        wav = torch.from_numpy(audio.T)

        result = predictor.forward([{"path": wav, "sample_rate": sr}])[0]
        row["PQ"] = result.get("PQ", "")
        row["CE"] = result.get("CE", "")
        row["CU"] = result.get("CU", "")

        print(f"[{i+1}/{len(rows)}] {track}/{stem}: PQ={row['PQ']} CE={row['CE']} CU={row['CU']}")
    except Exception as e:
        print(f"[{i+1}/{len(rows)}] BŁĄD {track}/{stem}: {e}")
        continue

    with open(RESULTS_CSV, "w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=rows[0].keys())
        writer.writeheader()
        writer.writerows(rows)

print("\n✅ Gotowe. PQ/CE/CU doliczone.")
```

```python
import pandas as pd

RESULTS_CSV = "/content/drive/MyDrive/Spleeter_Benchmark/results.csv"
SUMMARY_CSV = "/content/drive/MyDrive/Spleeter_Benchmark/summary_results.csv"

df = pd.read_csv(RESULTS_CSV)
for col in ["SDR", "SI-SDR", "PQ", "CE", "CU"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

summary = df.groupby("stem").agg({
    "SDR": ["mean", "median", "std"],
    "SI-SDR": ["mean", "median", "std"],
    "PQ": ["mean", "median", "std"],
    "CE": ["mean", "median", "std"],
    "CU": ["mean", "median", "std"],
}).round(4)

summary.to_csv(SUMMARY_CSV)
print(summary)
```

```python
import os
import shutil

OUTPUTS_DIR = "/content/drive/MyDrive/Spleeter_Benchmark/outputs"
AUDIO_DIR = "/content/drive/MyDrive/Spleeter_Benchmark/audio"
os.makedirs(AUDIO_DIR, exist_ok=True)

selected = sorted(os.listdir(OUTPUTS_DIR))[:5]
print("Wybrane utwory:", selected)

for i, track in enumerate(selected):
    src = os.path.join(OUTPUTS_DIR, track)
    dst = os.path.join(AUDIO_DIR, f"track_{i+1:03d}")
    os.makedirs(dst, exist_ok=True)
    for stem in ["vocals.wav", "drums.wav", "bass.wav", "other.wav"]:
        src_file = os.path.join(src, stem)
        if os.path.exists(src_file):
            shutil.copy(src_file, os.path.join(dst, stem))
    print(f"Skopiowano {track} -> track_{i+1:03d}")
```
